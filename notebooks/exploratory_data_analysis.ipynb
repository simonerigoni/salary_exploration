```python
import requests
from bs4 import BeautifulSoup

url = "https://www.payscale.com/research/IT/Job=Data_Engineer/Salary"
response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")

# Example: Extract salary information (adjust selectors based on page structure)
salary_data = soup.find_all(
    "span", class_="salary"
)  # Placeholder selector; inspect the page for actual classes
for data in salary_data:
    print(data.text)
```
